```python
import numpy as np
import pandas as pd
import torch

path = "TRAIN2.xlsx"
df = pd.read_excel(path)

# Chuyển đổi sang tensor
X = torch.tensor(df["midterm"].values, dtype=torch.float32).reshape(-1, 1)
Y = torch.tensor(df["final"].values, dtype=torch.float32).reshape(-1, 1)

# Khởi tạo tham số ngẫu nhiên thay vì bằng 0 để học nhanh hơn
W = torch.randn((1, 1), requires_grad=True)
b = torch.randn(1, requires_grad=True)

# Hyperparameters
epochs = 2000
learning_rate = 0.001 # Giảm lr nếu không chuẩn hóa dữ liệu

for epoch in range(epochs):
    # Forward pass
    Y_pred = X @ W + b

    # Compute Loss (MSE)
    loss = torch.mean((Y - Y_pred)**2)

    # Backward pass
    loss.backward()

    # Update weights
    with torch.no_grad():
        W -= learning_rate * W.grad
        b -= learning_rate * b.grad

    # Zero gradients
    W.grad.zero_()
    b.grad.zero_()

    if (epoch+1) % 500 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

print(f'\nKết quả: W = {W.item():.4f}, b = {b.item():.4f}')
```
